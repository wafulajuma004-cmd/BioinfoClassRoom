```python
#solution to integrated activity 1


1. # Raw sequence
raw_sequence = "   atggccaattggccnn "
2. # removing surrounding white space
clean_seq = raw_sequence.strip()
3. # converting to uppercase 
clean_seq= clean_seq.upper()
4. # printing cleaned sequence
print('Cleaned sequence:', clean_seq)
5. # printing the length of the cleaned sequence
length = len(clean_seq)
6. # printing A,T,G,C and T counts 
print('sequence length:',len(clean_seq))
print('A count:',clean_seq.count('A'))
print('T count:',clean_seq.count('T'))
print('G count:',clean_seq.count('G'))
print('C count:',clean_seq.count('C'))
print('N count:',clean_seq.count('N'))
7. # calculating the GC count using the clean sequence
gc_count = clean_seq.count('G') + clean_seq.count('C')
8.# calculating the percentage of the gc count to 2 decimal places 
gc_percentage = gc_count/length *100
print(f"GC percentage: {gc_percentage:.2f}%")


```
